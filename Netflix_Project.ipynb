```python
df['director'] = df['director'].fillna('No directors Listed')
df['cast'] = df['cast'].fillna('cast not Listed')
df['country'] = df['country'].fillna(' not Listed')
df['date_added'] = df['date_added'].fillna(' date_added not Listed')
df['rating'] = df['rating'].fillna('rating not Listed')
df['duration'] = df['duration'].fillna('duration not Listed')

df.head()
```

```python
#df.info() #no more NaN
```

```python
#Number of Netflix Movies Released per Year (Last 30 Years)

movies_df = df[(df['type'] == 'Movie') & (df['release_year'] >= 1996)]

yearly_counts = movies_df.groupby('release_year').size().reset_index(name='count')

plt.figure(figsize=(12, 6))
sns.lineplot(data=yearly_counts, x='release_year', y='count', marker='o', linewidth=2.5, color='crimson')

sns.lineplot()

plt.title('Number of Netflix Movies Released per Year (Last 30 Years)', fontsize=14, fontweight='bold')
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
#df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month

print(df['date_added'].info)
print(df['month_added'].info)
```

```python
tv_shows = df[df['type'] == 'TV Show']
monthly_tv_show_counts = tv_shows.groupby('month_added').size().reset_index(name='count')
print(monthly_tv_show_counts)
```

```python
import calendar
monthly_tv_show_counts['month_name'] = monthly_tv_show_counts['month_added'].apply(lambda x: calendar.month_abbr[int(x)])
print(monthly_tv_show_counts)
```

```python
sns.lineplot(data=monthly_tv_show_counts, x='month_name', y='count', marker='o', linewidth=2.5, color='crimson')
```

```python
# Highest number of TV Shows are released in July,Sept,Dec and least number of TV Shows are released in Jan,Feb,May
```

```python
filtered_directors_df = df[df['director'] != 'No directors Listed']

plt.figure(figsize=(18, 6))
ax = sns.countplot(data=filtered_directors_df, x='director', order=filtered_directors_df['director'].value_counts().index[:30], palette='viridis')
plt.title('Top 30 Directors with Most Movies & Tv Shows on Netflix', fontsize=16)
plt.xlabel('Director', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)
plt.xticks(rotation=90)

# Add count labels on top of each bar
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python
#Top 50 Most Frequent Actors / Cast Members on Netflix
df_exp=df.copy()
df_exp.drop(df_exp[df_exp['cast'] == 'cast not Listed'].index, inplace = True)
df_exp=df_exp.assign(cast=df_exp['cast'].str.split(',')).explode('cast').reset_index(drop=True)
count_val = df_exp['cast'].value_counts()
print(count_val[count_val.values <= 2].index.to_series().count())
dir = df_exp['cast'].value_counts().head(50).index

plt.figure(figsize=(18,6))
plot = sns.countplot(data=df_exp, x = 'cast', order = dir, hue= 'cast')
plot.set_xticklabels(labels=dir, rotation=90)
plt.title('Top 50 Most Frequent Actors / Cast Members on Netflix', fontsize=16)
plt.xlabel('Cast Members', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)
for p in plot.patches:
 plot.annotate('{:}'.format(p.get_height()), (p.get_x()+0.25, p.get_height()+0.01))
plt.show()
```
